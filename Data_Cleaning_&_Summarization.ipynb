```python
# The next line of code was used to drop rows where specficially diabetes has missing data

df_clean = df.dropna(subset=['has_diabetes']).copy()

# The next line of code was used to check how many of these rows you no longer have in your data

print(f"Started with {len(df)} rows")
print(f"Now have {len(df_clean)} rows")
```

```python
# The next line of code was used to fill any systolic BP using mean

mean_bp = df_clean['systolic_bp'].mean()
df_clean['systolic_bp'] = df_clean['systolic_bp'].fillna(mean_bp)
```

```python
# The next line of code was used to make sure the code from above worked

df_clean['systolic_bp'].isna().sum()
```

Note: The next lines of code will deal with transforming variables within our data

```python
# This next line of code will be used to creat BMI categories

df['bmi_category'] = pd.cut(df['bmi'],
                            bins=[0, 18.5, 25, 30, 100],
                            labels=['Underweight', 'Normal', 'Overweight', 'Obese'])
```

```python
# The next line of code was use to check our work from above

df['bmi_category'].value_counts()
```

```python
# The next line of code was used to show all the values that appear in the education column

df['education'].unique()
```

```python
# The next line of code was used to create a map dictionary, this is adjusted based on what is shown in your data dictionary depending on dataset

education_map = {
    1.0: 'less than 9th grade',
    2.0: '9-11th grade',
    3.0: 'High school grad',
    4.0: 'Some college or AA degree',
    5.0: 'College graduate',
    9.0: 'Do not know'
}

# The next line of code was used to apply the mapping we did above

df['education_label'] = df['education'].map(education_map)
```

```python
# This next line of code was used to make sure our work from above worked properly

df['education_label'].value_counts()
```

Note: The next lines of code are related to data summarization

```python
# The next line of code was used to calculate mean

df['systolic_bp'].mean()

# The next line of case was used to calculate standard deviation

df['systolic_bp'].mean()

# The next line of code was use to obtain both mean and standard deviation
df['systolic_bp'].describe()
```

```python
# The next line of code was used to get the median

df['age'].median()
```

```python
# The next line of code was used to find the IQR for only the 25th and 75th percentiles

df['age'].quantile([0.25, 0.75])
```

```python
# The next line of code was used to obtain IQR at all percentiles and both mean + SD

df['age'].describe()
```

```python
# The next line of code was used to obtain the mode of the data

df['age'].mode()
```

```python
# The next line of code was used to create a cross-tab of our dataset

pd.crosstab(df['has_diabetes'], df['bmi_category'])
```
