```python
# The next lines of code deal with IF/conditionals

# The next code was used to give a systolic BP limit for high and then for normal

systolic_bp = 120

if systolic_bp > 120:
  print ("High")
else:
    print ("Normal")
```

```python
def add_one(number):
  return number + 1

# The next line of code will add 3 to the function we made
add_one(3)

# The next line of code will be used to check systolic BP
def bp_check(systolic_bp):
  if systolic_bp > 120:
    bp_result = "High"
  else:
    bp_result = "Normal"
  return bp_result

# The next line of code was used to check the one above for bp checks
bp_check(150)
```

```python
# The next lines of code were used for Loops

# The next code was used to list different bps

bp_list = [120, 135, 170, 110, 100]

# The next code was used to check bps in the list above

for bp in bp_list:
  print(bp_check(bp))


# The next lines of code were used to define BMI and its categories

def bmi_check(bmi):
  if bmi < 18.5:
    bmi_result = "Underweight"
  elif bmi < 25:
    bmi_result = "Normal"
  elif bmi < 30:
    bmi_result = "Overweight"
  else:
    bmi_result = "Obese"
  return bmi_result


  # The next line of code was used to check bmi




```

```python
 # The next line of code was used to check bmi

bmi_check(24.9)
```

```python
# The next line of code was used to calculate BMI for group 2

def calculate_bmi(height, weight):
  bmi = weight/height ** 2
  return bmi

# The next codes were used to calculate bmi for the given values in 2 different ways

print(calculate_bmi(1.2, 57))

print(calculate_bmi(height = 1.2, weight = 57))
```
